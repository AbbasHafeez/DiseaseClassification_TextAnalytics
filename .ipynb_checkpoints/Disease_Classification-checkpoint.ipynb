```
# =====================
#    IMPORT LIBRARIES
# =====================
import pandas as pd
import numpy as np
import ast  # for safely evaluating stringified Python lists
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import PCA, TruncatedSVD
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.model_selection import KFold
from sklearn.preprocessing import LabelEncoder
import matplotlib.pyplot as plt
from tqdm import tqdm
from scipy.spatial.distance import cdist
from sklearn.metrics.pairwise import cosine_distances
from scipy.sparse import hstack

# ==========================
#       TASK 1
#   TF-IDF Feature Extraction
# ==========================

# 1. LOAD DATA
# Provide the correct path to your CSV files
disease_df = pd.read_csv("disease_features.csv")      # Contains textual columns
one_hot_df = pd.read_csv("encoded_output2.csv")       # Provided one-hot encoded matrix

# Inspect the first few rows
print("Disease Features (Textual) DataFrame HEAD:")
print(disease_df.head(), "\n")

print("One-hot Encoded DataFrame HEAD:")
print(one_hot_df.head(), "\n")

# 2. PARSE Risk Factors, Symptoms, Signs, and Subtype FROM STRINGIFIED LISTS
#    Then convert them to single string format

def parse_string_list(str_list):
    """
    Safely evaluate a string that looks like a Python list (e.g., "['fever','stress']").
    Returns a Python list.
    """
    try:
        return ast.literal_eval(str_list)
    except:
        # if parsing fails, return empty list
        return []

# Assuming your dataset columns are named "Risk Factors", "Symptoms", "Signs", "Subtype"
text_columns = ["Risk Factors", "Symptoms", "Signs"]  # Customize as needed
for col in text_columns:
    disease_df[col] = disease_df[col].apply(parse_string_list)

# 3. CONVERT EACH LIST INTO A SINGLE STRING
for col in text_columns:
    disease_df[col] = disease_df[col].apply(lambda x: " ".join(x))

# 4. APPLY TF-IDF VECTORIZATION TO EACH COLUMN (Risk Factors, Symptoms, Signs) SEPARATELY
tfidf_vectors = []
vectorizers = {}
for col in text_columns:
    vectorizer = TfidfVectorizer()
    vec = vectorizer.fit_transform(disease_df[col])
    tfidf_vectors.append(vec)
    vectorizers[col] = vectorizer

# 5. COMBINE THE TF-IDF MATRICES INTO A SINGLE FEATURE MATRIX
tfidf_feature_matrix = hstack(tfidf_vectors)

print("TF-IDF Combined Feature Matrix shape:", tfidf_feature_matrix.shape)

# 6. Compare the TF-IDF matrix with the provided one-hot encoded matrix (e.g., sparsity, unique features)
tfidf_nnz = tfidf_feature_matrix.nnz  # number of non-zero entries in TF-IDF
one_hot_nnz = one_hot_df.iloc[:, 1:].to_numpy().sum()  # assume first col is Disease name/category

tfidf_sparsity = 1 - (tfidf_nnz / (tfidf_feature_matrix.shape[0]*tfidf_feature_matrix.shape[1]))
one_hot_data = one_hot_df.iloc[:, 1:].values
one_hot_sparsity = 1 - (one_hot_nnz / (one_hot_data.shape[0]*one_hot_data.shape[1]))

print(f"TF-IDF matrix sparsity: {tfidf_sparsity:.4f}")
print(f"One-hot matrix sparsity: {one_hot_sparsity:.4f}")

# Count unique features
total_tfidf_features = sum(len(vectorizers[col].vocabulary_) for col in text_columns)
one_hot_features = one_hot_data.shape[1]

print(f"Total unique TF-IDF features (across 3 columns): {total_tfidf_features}")
print(f"Total one-hot features: {one_hot_features}")

# ==========================
#       TASK 2
#   Dimensionality Reduction
# ==========================

# Get disease labels (or categories).
# We'll assume there's a column called "Disease" or "Category" in disease_df. Adjust as needed.
if "Disease" in disease_df.columns:
    disease_labels = disease_df["Disease"].values
else:
    # or if you have a "Subtype" or "Category" column, use that
    disease_labels = disease_df["Subtype"].values

# Convert them to numeric labels for plotting colors
label_encoder = LabelEncoder()
encoded_labels = label_encoder.fit_transform(disease_labels)

# PCA and Truncated SVD on both TF-IDF and one-hot
NUM_COMPONENTS = 2  # for 2D visualization

# Convert TF-IDF to dense array for PCA
tfidf_array = tfidf_feature_matrix.toarray()

# ---- PCA ----
pca_tfidf = PCA(n_components=NUM_COMPONENTS, random_state=42)
tfidf_pca_2d = pca_tfidf.fit_transform(tfidf_array)

pca_one_hot = PCA(n_components=NUM_COMPONENTS, random_state=42)
one_hot_array = one_hot_data
one_hot_pca_2d = pca_one_hot.fit_transform(one_hot_array)

print("\nExplained variance ratio (TF-IDF, PCA):", pca_tfidf.explained_variance_ratio_)
print("Explained variance ratio (One-hot, PCA):", pca_one_hot.explained_variance_ratio_)

# ---- Truncated SVD ----
svd_tfidf = TruncatedSVD(n_components=NUM_COMPONENTS, random_state=42)
tfidf_svd_2d = svd_tfidf.fit_transform(tfidf_feature_matrix)  # can directly pass sparse

svd_one_hot = TruncatedSVD(n_components=NUM_COMPONENTS, random_state=42)
one_hot_svd_2d = svd_one_hot.fit_transform(one_hot_array)

print("\nExplained variance ratio (TF-IDF, SVD):", svd_tfidf.explained_variance_ratio_)
print("Explained variance ratio (One-hot, SVD):", svd_one_hot.explained_variance_ratio_)

# ---- VISUALIZATION (2D) PCA ----
plt.figure()
plt.scatter(tfidf_pca_2d[:, 0], tfidf_pca_2d[:, 1], c=encoded_labels)
plt.title("PCA (TF-IDF) - 2D")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.show()

plt.figure()
plt.scatter(one_hot_pca_2d[:, 0], one_hot_pca_2d[:, 1], c=encoded_labels)
plt.title("PCA (One-hot) - 2D")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.show()

# ---- VISUALIZATION (2D) SVD ----
plt.figure()
plt.scatter(tfidf_svd_2d[:, 0], tfidf_svd_2d[:, 1], c=encoded_labels)
plt.title("SVD (TF-IDF) - 2D")
plt.xlabel("Component 1")
plt.ylabel("Component 2")
plt.show()

plt.figure()
plt.scatter(one_hot_svd_2d[:, 0], one_hot_svd_2d[:, 1], c=encoded_labels)
plt.title("SVD (One-hot) - 2D")
plt.xlabel("Component 1")
plt.ylabel("Component 2")
plt.show()

# ==========================
#       TASK 3
#  KNN & Logistic Regression
# ==========================

def cross_validate_knn(
    X, 
    y, 
    k_values=[3,5,7], 
    distance_metrics=["euclidean", "manhattan", "cosine"], 
    folds=5
):
    """
    Perform KNN with given k-values and distance metrics using K-fold cross-validation.
    Returns a DataFrame with averaged performance metrics.
    """
    results = []
    kf = KFold(n_splits=folds, shuffle=True, random_state=42)
    
    for k in k_values:
        for metric in distance_metrics:
            accuracies = []
            precisions = []
            recalls = []
            f1s = []
            
            for train_idx, test_idx in kf.split(X):
                X_train, X_test = X[train_idx], X[test_idx]
                y_train, y_test = y[train_idx], y[test_idx]
                
                # define KNN
                if metric in ["euclidean", "manhattan"]:
                    knn = KNeighborsClassifier(n_neighbors=k, metric=metric)
                elif metric == "cosine":
                    # scikit-learn supports metric="cosine" in some versions
                    knn = KNeighborsClassifier(n_neighbors=k, metric="cosine")
                else:
                    # default
                    knn = KNeighborsClassifier(n_neighbors=k)
                
                knn.fit(X_train, y_train)
                y_pred = knn.predict(X_test)
                
                accuracies.append(accuracy_score(y_test, y_pred))
                precisions.append(precision_score(y_test, y_pred, average='macro'))
                recalls.append(recall_score(y_test, y_pred, average='macro'))
                f1s.append(f1_score(y_test, y_pred, average='macro'))
            
            # summarize
            results.append({
                "K": k,
                "Metric": metric,
                "Accuracy": np.mean(accuracies),
                "Precision": np.mean(precisions),
                "Recall": np.mean(recalls),
                "F1-score": np.mean(f1s)
            })
    return pd.DataFrame(results)

def cross_validate_logistic(X, y, folds=5):
    """
    5-fold cross-validation with Logistic Regression.
    Returns average accuracy and F1 score.
    """
    kf = KFold(n_splits=folds, shuffle=True, random_state=42)
    accuracies = []
    f1s = []
    
    for train_idx, test_idx in kf.split(X):
        X_train, X_test = X[train_idx], X[test_idx]
        y_train, y_test = y[train_idx], y[test_idx]
        
        clf = LogisticRegression(max_iter=1000)
        clf.fit(X_train, y_train)
        y_pred = clf.predict(X_test)
        
        accuracies.append(accuracy_score(y_test, y_pred))
        f1s.append(f1_score(y_test, y_pred, average='macro'))
    
    return np.mean(accuracies), np.mean(f1s)

# Convert label arrays if not numeric
if disease_labels.dtype.kind not in ['i', 'u', 'f']:
    disease_labels = label_encoder.fit_transform(disease_labels)

# A) KNN on TF-IDF
print("\nKNN on TF-IDF Features")
tfidf_array = tfidf_feature_matrix.toarray()
knn_tfidf_results = cross_validate_knn(tfidf_array, disease_labels)
print(knn_tfidf_results)

# B) KNN on One-hot
print("\nKNN on One-hot Features")
knn_one_hot_results = cross_validate_knn(one_hot_array, disease_labels)
print(knn_one_hot_results)

# C) Logistic Regression on TF-IDF
print("\nLogistic Regression on TF-IDF:")
lr_tfidf_acc, lr_tfidf_f1 = cross_validate_logistic(tfidf_array, disease_labels)
print(f"Accuracy: {lr_tfidf_acc:.3f}, F1-score: {lr_tfidf_f1:.3f}")

# D) Logistic Regression on One-hot
print("\nLogistic Regression on One-hot:")
lr_onehot_acc, lr_onehot_f1 = cross_validate_logistic(one_hot_array, disease_labels)
print(f"Accuracy: {lr_onehot_acc:.3f}, F1-score: {lr_onehot_f1:.3f}")

# ==========================
#       TASK 4
#   Critical Analysis
# ==========================
analysis_text = """
===============================
      CRITICAL ANALYSIS
===============================

1. Why TF-IDF might outperform one-hot encoding (or vice versa) for this dataset?
   TF-IDF captures the relative importance of each word across documents, reducing
   the weight of common but less informative terms. This could help distinguish diseases
   by unique terms better than a one-hot approach, which treats all terms equally.
   However, if the dataset is small or each feature is highly specific, one-hot might suffice
   or even be better in certain edge cases.

2. Clinical relevance of results
   If we see that clusters formed by TF-IDF reflect actual disease categories (e.g.,
   "cardiovascular," "neurological"), it implies that these textual features correlate strongly
   with real-world distinctions. If one-hot or TF-IDF lumps together certain diseases incorrectly,
   we should investigate how their textual descriptions might overlap or fail to differentiate them.

3. Limitations of both encoding methods
   - TF-IDF still depends on bag-of-words, ignoring word order and context.
   - One-hot can become very large and sparse, especially for textual data with many unique tokens.
   - Both methods ignore semantic similarity between tokens (e.g., "heart attack" vs.
     "myocardial infarction").

Overall, the choice might depend on how well each representation captures clinically relevant
distinctions among diseases.
"""
print(analysis_text)
```
